```python
import os
import cv2
import numpy as np
import random
from collections import Counter
import matplotlib.pyplot as plt
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from tensorflow.keras.utils import to_categorical
import kagglehub # Import kagglehub

train_path = 'chest_xray/train'
test_path = 'chest_xray/test'
img_size = 224
categories = ['NORMAL', 'PNEUMONIA']
print("Path: ", train_path)

# Function to load images and labels from a given path
def load_data(folder_path, base_path):  # Added base_path as an argument
    data = []
    labels = []
    file_paths = {category: [] for category in categories}

    for category in categories:
        class_label = categories.index(category)
        print(folder_path)
        # Use os.path.join to combine the base path with the folder path and category
        category_path = os.path.join(base_path, folder_path, category)  # Use base_path
        print(category_path)

        for img in os.listdir(category_path):
            img_path = os.path.join(category_path, img)
            file_paths[category].append(img_path)

    return file_paths

#Download the dataset and assign to path here
path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
print("Path to dataset files:", path)

# Load file paths
train_file_paths = load_data(train_path, path)  # Pass path to load_data

# Find the class with the minimum images
min_count = min(len(train_file_paths[c]) for c in categories)

# Downsample the larger class to match the smaller class
balanced_file_paths = {c: random.sample(train_file_paths[c], min_count) for c in categories}

# Load balanced dataset
X_train, y_train = [], []

for category in categories:
    class_label = categories.index(category)
    for img_path in balanced_file_paths[category]:
        img_array = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img_array is not None:
            resized_img = cv2.resize(img_array, (img_size, img_size))
            X_train.append(resized_img)
            y_train.append(class_label)

# Convert to numpy arrays
X_train = np.array(X_train).reshape(-1, img_size, img_size, 1) / 255.0  # Normalize
y_train = np.array(y_train)
y_train = to_categorical(y_train, num_classes=len(categories))

# Image Augmentation
data_gen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.2,
    height_shift_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)

# Expand dataset using augmentation
X_train_augmented = []
y_train_augmented = []

for i in range(len(X_train)):
    img = X_train[i]
    img = np.expand_dims(img, axis=0)  # Reshape for augmentation
    it = data_gen.flow(img, batch_size=1)
    for _ in range(3):  # Generate 3 augmented images per original
        # aug_img = it.next()[0]  <-- Replace this line with the following:
        aug_img = next(it)[0] # Use next(it) to get the next item from the iterator
        X_train_augmented.append(aug_img)
        y_train_augmented.append(y_train[i])


X_train_augmented = np.array(X_train_augmented)
y_train_augmented = np.array(y_train_augmented)

# Define CNN Model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(img_size, img_size, 1)),
    MaxPooling2D(2, 2),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(len(categories), activation='softmax')
])

# Compile the model
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])

# Train the model
history = model.fit(X_train_augmented, y_train_augmented, epochs=10, batch_size=32, validation_split=0.2)

# Visualization of augmented images
def show_augmented_images(X, y, categories):
    fig, axes = plt.subplots(1, len(categories), figsize=(10, 5))
    for i, category in enumerate(categories):
        idx = np.where(y[:, i] == 1)[0][5]  # Get first occurrence of class
        axes[i].imshow(X[idx].squeeze(), cmap='gray')
        axes[i].set_title(f"Augmented {category}")
        axes[i].axis('off')
    plt.show()

show_augmented_images(X_train_augmented, y_train_augmented, categories)

# Analyze dataset distribution
train_image_dims = []
train_image_formats = []

for category in categories:
    category_path = os.path.join(train_path, category) # Remove `path` here, analyze downloaded path for dimensions
    for img in os.listdir(os.path.join(path, category_path)): # Add `path` when listing downloaded files
        img_path = os.path.join(path, category_path, img) # Add `path` for reading the downloaded image
        img_array = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
        if img_array is not None:
            train_image_dims.append(img_array.shape)
            train_image_formats.append(img.split('.')[-1])

unique_dims = Counter(train_image_dims)
unique_formats = Counter(train_image_formats)

print("Unique image dimensions before resizing:", unique_dims)
print("Image format distribution:", unique_formats)

# Plot new class distribution
class_counts = Counter(np.argmax(y_train_augmented, axis=1))

plt.bar(categories, [class_counts[i] for i in range(len(categories))], color=['blue', 'red'])
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.title("Augmented Class Distribution")
plt.show()
```

```python
import kagglehub
import os
import cv2
import numpy as np
import random
from collections import Counter
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential, Model
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout, GlobalAveragePooling2D
from tensorflow.keras.applications import VGG16, ResNet50
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split

path = kagglehub.dataset_download("paultimothymooney/chest-xray-pneumonia")
print("Path to dataset files:", path)

# Define Paths and Constants
train_path = os.path.join(path, 'chest_xray/train') # Update train_path to include path from kagglehub
test_path = os.path.join(path, 'chest_xray/test')  # Update test_path to include path from kagglehub
img_size = 224
categories = ['NORMAL', 'PNEUMONIA']
num_classes = len(categories)

# Function to load and balance file paths
def load_balanced_file_paths(folder_path):
    file_paths = {category: [] for category in categories}

    for category in categories:
        class_label = categories.index(category)
        category_path = os.path.join(folder_path, category)

        for img in os.listdir(category_path):
            img_path = os.path.join(category_path, img)
            file_paths[category].append(img_path)

    # Find the class with the minimum images
    min_count = min(len(file_paths[c]) for c in categories)

    # Downsample the larger class to match the smaller class
    balanced_file_paths = {c: random.sample(file_paths[c], min_count) for c in categories}
    return balanced_file_paths

# Load balanced file paths for training data
balanced_file_paths = load_balanced_file_paths(train_path)  # Call the function here


# Load balanced dataset
X_train, y_train = [], []

for category in categories:
    class_label = categories.index(category)
    for img_path in balanced_file_paths[category]:
        # Load images in color (3 channels)
        img_array = cv2.imread(img_path, cv2.IMREAD_COLOR)  # Change to cv2.IMREAD_COLOR
        if img_array is not None:
            resized_img = cv2.resize(img_array, (img_size, img_size))
            X_train.append(resized_img)
            y_train.append(class_label)

# Convert to numpy arrays and normalize
X_train = np.array(X_train) / 255.0  # Normalize after converting to NumPy array
y_train = np.array(y_train)
# Load Data
X_train, X_val, y_train, y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
# Convert labels to categorical (one-hot encoding)
y_train = to_categorical(y_train, num_classes)
y_val = to_categorical(y_val, num_classes)

# Transfer Learning - Load Pretrained Model
base_model = VGG16(weights='imagenet', include_top=False, input_shape=(img_size, img_size, 3))
for layer in base_model.layers[:-4]:
    layer.trainable = False  # Freeze all layers except last 4

# Custom Layers
x = base_model.output
x = GlobalAveragePooling2D()(x)
x = Dense(256, activation='relu')(x)
x = Dropout(0.5)(x)
output_layer = Dense(num_classes, activation='softmax')(x)

# Fine-Tuned Model
model_transfer = Model(inputs=base_model.input, outputs=output_layer)
model_transfer.compile(optimizer=Adam(learning_rate=0.0001), loss='categorical_crossentropy', metrics=['accuracy'])

# Train the Model
history_transfer = model_transfer.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_val, y_val))

# Basic CNN Model for Comparison
model_cnn = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(img_size, img_size, 3)),
    MaxPooling2D(2, 2),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),
    Dense(num_classes, activation='softmax')
])

model_cnn.compile(optimizer=Adam(), loss='categorical_crossentropy', metrics=['accuracy'])

# Train Basic CNN Model
history_cnn = model_cnn.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_val, y_val))

# Compare Performance
plt.plot(history_cnn.history['accuracy'], label='Basic CNN Accuracy')
plt.plot(history_transfer.history['accuracy'], label='Transfer Learning Accuracy')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.legend()
plt.show()

# Handle Class Imbalance
# Handle Class Imbalance (Fixed Calculation)
class_counts = Counter(np.argmax(y_train, axis=1))  # Convert one-hot back to class labels
max_count = max(class_counts.values())
class_weights = {cls: max_count / count for cls, count in class_counts.items()}

# Train Again with Class Weights
history_transfer_weighted = model_transfer.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_val, y_val), class_weight=class_weights)
```
